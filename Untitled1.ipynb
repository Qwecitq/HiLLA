```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from scipy.signal import detrend
from scipy.fft import fft2, fftfreq, fftshift
from scipy.ndimage import gaussian_filter

# ============================================================
# ASSUMPTION:
# data(time, lat, lon)
# clim(dayofyear, lat, lon)
# ============================================================

# ------------------------------------------------------------
# 1. Remove climatology
# ------------------------------------------------------------
anom = ukesm_hist_ua.sel(plev=85000)['ua'].groupby('time.dayofyear') - clim_850
# anom.load()
# Optional detrend
# anom_vals = detrend(anom.values, axis=0)
# anom = xr.DataArray(anom_vals, coords=anom.coords, dims=anom.dims)
# Store NaN mask before filling
nan_mask = np.isnan(anom.values)

# Fill NaNs with 0 for detrending
anom_filled = anom.fillna(0.0)
anom = anom_filled
# ------------------------------------------------------------
# 2. Restrict to tropics
# ------------------------------------------------------------
anom = anom.sel(lat=slice(-15, 15))

# Ensure symmetric latitude grid
lat = anom['lat'].values
assert np.allclose(lat, -lat[::-1]), "Latitude grid must be symmetric about equator"

# ------------------------------------------------------------
# 3. Symmetric / antisymmetric decomposition
# ------------------------------------------------------------
north = anom.sel(lat=slice(0, 15))
south = anom.sel(lat=slice(-15, 0)).isel(lat=slice(None, None, -1))

sym  = 0.5 * (north.values + south.values)
asym = 0.5 * (north.values - south.values)

# Average over latitude AFTER decomposition
sym  = sym.mean(axis=1)   # shape: (time, lon)
asym = asym.mean(axis=1)  # shape: (time, lon)

# ------------------------------------------------------------
# 4. Segment data
# ------------------------------------------------------------
def segment_data(data, window=96, overlap=60):
    step = window - overlap
    segments = []
    for start in range(0, data.shape[0] - window + 1, step):
        segments.append(data[start:start + window, :])
    return np.array(segments)

# ------------------------------------------------------------
# 5. Apply taper — FIXED syntax error
# ------------------------------------------------------------
def apply_taper(seg):
    nt, nx = seg.shape
    time_win = np.hanning(nt)
    lon_win  = np.hanning(nx)
    return seg * time_win[:, None] * lon_win[None, :]

# ------------------------------------------------------------
# 6. Compute power spectrum
# ------------------------------------------------------------
def compute_power(data_array):
    segments = segment_data(data_array)
    power_list = []
    for seg in segments:
        seg = apply_taper(seg)
        F = fft2(seg, axes=(0, 1))
        power = np.abs(F) ** 2
        power_list.append(power)
    return np.mean(power_list, axis=0)

power_sym  = compute_power(sym)
power_asym = compute_power(asym)

# ------------------------------------------------------------
# 7. Frequency and wavenumber axes
# ------------------------------------------------------------
nt, nx = power_sym.shape
freq = fftfreq(nt, d=1.0)       # cycles/day
k    = fftfreq(nx, d=1.0) * nx  # zonal wavenumber

# Shift zero to center
freq       = fftshift(freq)
k          = fftshift(k)
power_sym  = fftshift(power_sym,  axes=(0, 1))
power_asym = fftshift(power_asym, axes=(0, 1))

# Convert frequency to period
with np.errstate(divide='ignore', invalid='ignore'):
    period = np.where(freq != 0, 1.0 / np.abs(freq), np.nan)

# ------------------------------------------------------------
# 8. Background spectrum
# ------------------------------------------------------------
def background_spectrum(power):
    return gaussian_filter(power, sigma=[8, 3])

bg_sym  = background_spectrum(power_sym)
bg_asym = background_spectrum(power_asym)

norm_sym  = power_sym  / bg_sym
norm_asym = power_asym / bg_asym

# ------------------------------------------------------------
# 9. Plotting
# ------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 7), sharey=True)

wn_lim  = [-15, 15]
per_lim = [2, 96]

wn_idx  = (k      >= wn_lim[0])  & (k      <= wn_lim[1])
per_idx = (period >= per_lim[0]) & (period <= per_lim[1])

def plot_panel(ax, power, title):
    cf = ax.contourf(
        k[wn_idx],
        period[per_idx],
        power[np.ix_(per_idx, wn_idx)],
        levels=np.arange(0.5, 3.0, 0.1),
        cmap='RdBu_r',
        extend='both'
    )
    ax.set_yscale('log')
    ax.set_ylim(per_lim[1], per_lim[0])
    ax.set_xlabel('Zonal Wavenumber', fontsize=12)
    ax.set_ylabel('Period (days)',     fontsize=12)
    ax.set_title(title, fontsize=13)
    ax.axvline(0, color='k', linewidth=0.8)
    # Custom y-tick labels for readability
    yticks = [2, 3, 5, 7, 10, 14, 20, 30, 50, 96]
    ax.set_yticks(yticks)
    ax.set_yticklabels([str(y) for y in yticks])
    return cf

cf1 = plot_panel(axes[0], norm_sym,  "Symmetric Component")
cf2 = plot_panel(axes[1], norm_asym, "Antisymmetric Component")

# ------------------------------------------------------------
# 10. Dispersion curves — FIXED syntax errors
# ------------------------------------------------------------
g    = 9.81
a    = 6.371e6
beta = 2.28e-11

def omega_to_cpd(omega):
    return omega * 86400.0 / (2.0 * np.pi)  # FIXED

k_arr = np.arange(-15, 16, 0.1)
k_rad = k_arr / a

he_list = [12, 25, 50]
colors  = ['gray', 'black', 'dimgray']

for he, col in zip(he_list, colors):
    c = np.sqrt(g * he)

    # --- Kelvin wave (symmetric panel, eastward only) ---
    omega_kelvin  = c * k_rad
    freq_kelvin   = omega_to_cpd(omega_kelvin)
    period_kelvin = np.where(freq_kelvin > 0, 1.0 / freq_kelvin, np.nan)
    axes[0].plot(
        k_arr, period_kelvin,
        color=col, linewidth=0.9, linestyle='--',
        label=f'Kelvin he={he}m'
    )

    # --- n=1 Equatorial Rossby wave (antisymmetric panel) --- FIXED
    n = 1
    omega_er  = -(beta * k_rad) / (k_rad**2 + (2*n + 1) * beta / c)
    freq_er   = omega_to_cpd(omega_er)
    period_er = np.where(np.abs(freq_er) > 0, 1.0 / np.abs(freq_er), np.nan)
    axes[1].plot(
        k_arr, period_er,
        color=col, linewidth=0.9, linestyle='--',
        label=f'ER n=1 he={he}m'
    )

    # --- n=0 Mixed Rossby-Gravity wave (antisymmetric panel) ---
    # omega^2 - beta/k * omega - c^2 * k^2 = 0
    with np.errstate(invalid='ignore'):
        discriminant = (beta / k_rad)**2 + 4 * c**2 * k_rad**2
        omega_mrg = 0.5 * (beta / k_rad + np.sqrt(discriminant))
    freq_mrg   = omega_to_cpd(omega_mrg)
    period_mrg = np.where(np.abs(freq_mrg) > 0, 1.0 / np.abs(freq_mrg), np.nan)
    axes[1].plot(
        k_arr, period_mrg,
        color=col, linewidth=0.9, linestyle=':',
        label=f'MRG he={he}m'
    )

axes[0].legend(fontsize=7, loc='upper right')
axes[1].legend(fontsize=7, loc='upper right')

# ------------------------------------------------------------
# 11. MJO box (symmetric panel) — adjusted coordinates
# ------------------------------------------------------------
axes[0].add_patch(
    plt.Rectangle(
        (1, 30), 4, 66,           # wn 1-5, period 30-96 days
        fill=False,
        edgecolor='green',
        linestyle='--',
        linewidth=1.5,
        label='MJO'
    )
)
axes[0].text(1.2, 28, 'MJO', color='green', fontsize=9)

# ------------------------------------------------------------
# 12. Shared colorbar and final formatting
# ------------------------------------------------------------
fig.colorbar(cf1, ax=axes.ravel().tolist(), label='Normalized Power', shrink=0.8)
plt.suptitle('Wheeler-Kiladis Wavenumber-Frequency Diagram', fontsize=15, y=1.01)
plt.tight_layout()
plt.savefig('wk_diagram.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from scipy.fft import fft2, fftfreq, fftshift
from scipy.ndimage import gaussian_filter

# ============================================================
# Wheeler-Kiladis Wavenumber-Frequency Diagram
# INPUT ASSUMPTIONS:
#   ukesm_hist_ua : xarray Dataset (time, plev, lat, lon)
#   clim_850      : xarray DataArray (dayofyear, lat, lon)
# ============================================================

# ------------------------------------------------------------
# STEP 1: Remove climatology and fill NaNs
# ------------------------------------------------------------
print("=" * 60)
print("STEP 1: Removing climatology...")

raw = ukesm_hist_ua.sel(plev=85000)['ua']
print(f"  Raw data shape       : {raw.shape}")
print(f"  Raw NaN count        : {int(np.isnan(raw.values).sum())}")

anom = raw.groupby('time.dayofyear') - clim_850
print(f"  Anomaly NaN count    : {int(np.isnan(anom.values).sum())}")

# Fill NaNs with 0 before FFT
anom = anom.fillna(0.0)
# print(f"  After fillna NaN count: {int(np.isnan(anom.values).sum())}")
print("  STEP 1: DONE")

# ------------------------------------------------------------
# STEP 2: Restrict to tropics and sort latitudes
# ------------------------------------------------------------
print("=" * 60)
print("STEP 2: Restricting to tropical band (-15 to 15)...")

# Sort latitudes to ensure ascending order before slicing
anom = anom.sortby('lat')

# Handle both ascending and descending latitude ordering
lat_full = anom['lat'].values
if lat_full[0] > lat_full[-1]:
    anom = anom.sel(lat=slice(15, -15))
else:
    anom = anom.sel(lat=slice(-15, 15))

anom = anom.sortby('lat')
lat  = anom['lat'].values

print(f"  Selected latitudes   : {lat}")
print(f"  Anom shape           : {anom.shape}")
# print(f"  NaN count after sel  : {int(np.isnan(anom.values).sum())}")
print("  STEP 2: DONE")
```

```python
# ------------------------------------------------------------
# STEP 4: Segment data
# ------------------------------------------------------------
print("=" * 60)
print("STEP 4: Segmenting data...")

def segment_data(data, window=96, overlap=60):
    """
    Segment time series into overlapping windows.
    data   : (time, lon)
    window : segment length in days
    overlap: overlap between segments in days
    """
    step     = window - overlap
    segments = []
    for start in range(0, data.shape[0] - window + 1, step):
        segments.append(data[start:start + window, :])
    return np.array(segments)

# Test segmentation
test_segs = segment_data(sym)
print(f"  Number of segments   : {len(test_segs)}")
print(f"  Segment shape        : {test_segs[0].shape}")
# print(f"  NaN in segments      : {np.isnan(test_segs).sum()}")
print("  STEP 4: DONE")

# ------------------------------------------------------------
# STEP 5: Apply Hann taper
# ------------------------------------------------------------
print("=" * 60)
print("STEP 5: Defining taper function...")

def apply_taper(seg):
    """
    Apply 2D Hann taper to reduce spectral leakage.
    seg: (n_time, n_lon)
    """
    nt, nx   = seg.shape
    time_win = np.hanning(nt)
    lon_win  = np.hanning(nx)
    return seg * time_win[:, None] * lon_win[None, :]  # FIXED: * operator

print("  Taper function defined. STEP 5: DONE")
```

```python
# ------------------------------------------------------------
# STEP 6: Compute power spectrum
# ------------------------------------------------------------
print("=" * 60)
print("STEP 6: Computing power spectrum...")

def compute_power(data_array, window=96, overlap=60):
    """
    Compute ensemble-averaged 2D power spectrum.
    data_array: (time, lon) — must be NaN free
    """
    segments   = segment_data(data_array, window=window, overlap=overlap)
    power_list = []
    for i, seg in enumerate(segments):
        # Safety fill — should already be NaN free
        seg   = np.nan_to_num(seg, nan=0.0)
        seg   = apply_taper(seg)
        F     = fft2(seg, axes=(0, 1))
        power = np.abs(F) ** 2
        power_list.append(power)
    return np.mean(power_list, axis=0)

power_sym  = compute_power(sym)
power_asym = compute_power(asym)

print(f"  power_sym shape      : {power_sym.shape}")
# print(f"  power_sym NaN count  : {np.isnan(power_sym).sum()}")
print(f"  power_sym min/max    : {np.nanmin(power_sym):.4f} / {np.nanmax(power_sym):.4f}")
# print(f"  power_asym NaN count : {np.isnan(power_asym).sum()}")
print("  STEP 6: DONE")

# ------------------------------------------------------------
# STEP 7: Frequency and wavenumber axes
# ------------------------------------------------------------
print("=" * 60)
print("STEP 7: Building frequency and wavenumber axes...")

nt, nx = power_sym.shape
freq   = fftfreq(nt, d=1.0)       # cycles/day
k      = fftfreq(nx, d=1.0) * nx  # integer zonal wavenumber

# Shift zero frequency to center
freq       = fftshift(freq)
k          = fftshift(k)
power_sym  = fftshift(power_sym,  axes=(0, 1))
power_asym = fftshift(power_asym, axes=(0, 1))

# Convert frequency to period (days)
with np.errstate(divide='ignore', invalid='ignore'):
    period = np.where(freq != 0, 1.0 / np.abs(freq), np.nan)

print(f"  Frequency range      : {freq[freq > 0].min():.4f} to {freq.max():.4f} cpd")
print(f"  Wavenumber range     : {k.min()} to {k.max()}")
print(f"  Period range         : {np.nanmin(period):.1f} to {np.nanmax(period):.1f} days")
print("  STEP 7: DONE")

# ------------------------------------------------------------
# STEP 8: Background spectrum and normalization
# ------------------------------------------------------------
print("=" * 60)
print("STEP 8: Computing background spectrum and normalizing...")

def background_spectrum(power, sigma=(8, 3)):
    """
    Smooth power spectrum with anisotropic Gaussian filter
    to estimate the red-noise background.
    sigma: (freq_sigma, wn_sigma)
    """
    return gaussian_filter(power.astype(float), sigma=list(sigma))

bg_sym  = background_spectrum(power_sym)
bg_asym = background_spectrum(power_asym)

# Avoid division by zero
bg_sym  = np.where(bg_sym  == 0, np.nan, bg_sym)
bg_asym = np.where(bg_asym == 0, np.nan, bg_asym)

norm_sym  = power_sym  / bg_sym
norm_asym = power_asym / bg_asym

# print(f"  bg_sym NaN count     : {np.isnan(bg_sym).sum()}")
# print(f"  norm_sym NaN count   : {np.isnan(norm_sym).sum()}")
print(f"  norm_sym min/max     : {np.nanmin(norm_sym):.4f} / {np.nanmax(norm_sym):.4f}")
print(f"  norm_asym min/max    : {np.nanmin(norm_asym):.4f} / {np.nanmax(norm_asym):.4f}")
print("  STEP 8: DONE")
```

```python
# ------------------------------------------------------------
# STEP 9: Subset for plotting
# ------------------------------------------------------------
print("=" * 60)
print("STEP 9: Subsetting for plot domain...")

wn_lim  = [-15, 15]
per_lim = [2,   96]

wn_idx  = (k      >= wn_lim[0])  & (k      <= wn_lim[1])
per_idx = (period >= per_lim[0]) & (period <= per_lim[1])

k_plot      = k[wn_idx]
period_plot = period[per_idx]

# Sort period for monotonic y-axis
sort_idx     = np.argsort(period_plot)
period_plot  = period_plot[sort_idx]

norm_sym_plot  = norm_sym[np.ix_(per_idx,  wn_idx)][sort_idx, :]
norm_asym_plot = norm_asym[np.ix_(per_idx, wn_idx)][sort_idx, :]

print(f"  k_plot range         : {k_plot.min()} to {k_plot.max()}")
print(f"  period_plot range    : {np.nanmin(period_plot):.1f} to {np.nanmax(period_plot):.1f} days")
print(f"  norm_sym_plot shape  : {norm_sym_plot.shape}")
# print(f"  norm_sym_plot NaN    : {np.isnan(norm_sym_plot).sum()}")
print("  STEP 9: DONE")

# ------------------------------------------------------------
# STEP 10: Plotting
# ------------------------------------------------------------
print("=" * 60)
print("STEP 10: Plotting...")

fig, axes = plt.subplots(1, 2, figsize=(14, 7), sharey=True)

def plot_panel(ax, power, title):
    cf = ax.contourf(
        k_plot,
        period_plot,
        power,
        levels=np.arange(0.5, 3.0, 0.1),
        cmap='RdBu_r',
        extend='both'
    )
    ax.set_yscale('log')
    ax.set_ylim(per_lim[1], per_lim[0])
    ax.set_xlabel('Zonal Wavenumber', fontsize=12)
    ax.set_ylabel('Period (days)',     fontsize=12)
    ax.set_title(title, fontsize=13)
    ax.axvline(0, color='k', linewidth=0.8)
    yticks = [2, 3, 5, 7, 10, 14, 20, 30, 50, 96]
    ax.set_yticks(yticks)
    ax.set_yticklabels([str(y) for y in yticks])
    ax.grid(alpha=0.2)
    return cf

cf1 = plot_panel(axes[0], norm_sym_plot,  "Symmetric Component")
cf2 = plot_panel(axes[1], norm_asym_plot, "Antisymmetric Component")

print("  Contour panels plotted.")

# ------------------------------------------------------------
# STEP 11: Dispersion curves
# ------------------------------------------------------------
print("  Adding dispersion curves...")

g    = 9.81
a    = 6.371e6
beta = 2.28e-11

def omega_to_cpd(omega):
    return omega * 86400.0 / (2.0 * np.pi)   # FIXED: * operator

k_arr = np.arange(-15, 16, 0.1)
k_rad = k_arr / a

he_list = [12, 25, 50]
colors  = ['gray', 'black', 'dimgray']

for he, col in zip(he_list, colors):
    c = np.sqrt(g * he)

    # --- Kelvin wave (symmetric panel, eastward only) ---
    omega_kelvin  = c * k_rad                           # FIXED
    freq_kelvin   = omega_to_cpd(omega_kelvin)
    period_kelvin = np.where(
        freq_kelvin > 0, 1.0 / freq_kelvin, np.nan
    )
    axes[0].plot(
        k_arr, period_kelvin,
        color=col, linewidth=0.9, linestyle='--',
        label=f'Kelvin he={he}m'
    )

    # --- n=1 Equatorial Rossby wave (symmetric panel) ---
    n        = 1
    omega_er = -(beta * k_rad) / (           # FIXED
        k_rad**2 + (2*n + 1) * beta / c      # FIXED
    )
    freq_er   = omega_to_cpd(omega_er)
    period_er = np.where(
        np.abs(freq_er) > 0, 1.0 / np.abs(freq_er), np.nan
    )
    axes[0].plot(
        k_arr, period_er,
        color=col, linewidth=0.9, linestyle='-.',
        label=f'ER n=1 he={he}m'
    )

    # --- n=0 Mixed Rossby-Gravity wave (antisymmetric panel) ---
    with np.errstate(invalid='ignore', divide='ignore'):
        discriminant = (beta / k_rad)**2 + 4 * c**2 * k_rad**2  # FIXED
        omega_mrg    = 0.5 * (beta / k_rad + np.sqrt(discriminant))
    freq_mrg   = omega_to_cpd(omega_mrg)
    period_mrg = np.where(
        np.abs(freq_mrg) > 0, 1.0 / np.abs(freq_mrg), np.nan
    )
    axes[1].plot(
        k_arr, period_mrg,
        color=col, linewidth=0.9, linestyle=':',
        label=f'MRG he={he}m'
    )

    # --- n=1 ER wave on antisymmetric panel as well ---
    axes[1].plot(
        k_arr, period_er,
        color=col, linewidth=0.9, linestyle='--',
        label=f'ER n=1 he={he}m'
    )

axes[0].legend(fontsize=7, loc='upper right')
axes[1].legend(fontsize=7, loc='upper right')

# ------------------------------------------------------------
# STEP 12: MJO box
# ------------------------------------------------------------
print("  Adding MJO box...")

from matplotlib.patches import Rectangle

axes[0].add_patch(
    Rectangle(
        (1, 30), 4, 66,
        fill=False,
        edgecolor='green',
        linestyle='--',
        linewidth=1.5
    )
)
axes[0].text(1.2, 28, 'MJO', color='green', fontsize=9)

# ------------------------------------------------------------
# STEP 13: Colorbar and save
# ------------------------------------------------------------
print("  Adding colorbar and saving...")

fig.colorbar(
    cf1,
    ax=axes.ravel().tolist(),
    label='Normalized Power',
    shrink=0.8
)
plt.suptitle(
    'Wheeler-Kiladis Wavenumber-Frequency Diagram',
    fontsize=15, y=1.01
)
plt.tight_layout()
plt.savefig('wk_diagram.png', dpi=300, bbox_inches='tight')
plt.show()

print("=" * 60)
print("ALL STEPS COMPLETE. Figure saved to wk_diagram.png")
print("=" * 60)
```
